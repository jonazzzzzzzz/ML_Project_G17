# <font size="6" color='#E8800A'>3: Abstract</font> <a class="anchor" id="resumo"></a>
<b><font size="5">Resumo do trabalho com contexto, objetivos, método, resultados e conclusões</font></b><br>
[Back to TOC](#toc)

**Contexto.** A Everything & Then Some, Ltd. é uma empresa grossista que fornece empresas e retalhistas especializados com um portfólio alargado de produtos. Os seus clientes têm padrões de compra muito diferentes, o que torna difícil antecipar quando voltam a comprar. A empresa pretende usar o histórico de vendas para apoiar ações comerciais mais direcionadas.

**Objetivo.** Este projeto desenvolve um modelo de classificação binária que prevê se um cliente fará nova compra nos 90 dias seguintes à transação atual. Para isso, trabalhámos com 6 534 ocasiões de compra históricas com o target conhecido (`train.csv`), 986 ocasiões posteriores sem target (`test.csv`) e uma tabela opcional de atributos dos clientes (`clientes.csv`). As previsões para o conjunto de teste foram submetidas numa competição Kaggle.

### Questão de negócio

> **Dado o cliente e a compra que acabou de fazer, voltará a comprar nos 90 dias seguintes?**

### Decisão que o modelo apoia

O modelo devolve, para cada compra, a indicação (e a probabilidade) de o cliente voltar ou não a comprar nos 90 dias seguintes. Essa informação serve para priorizar a atenção comercial, por exemplo:

- [ex.: concentrar contactos de acompanhamento nos clientes com baixa probabilidade de recompra]
- [ex.: distinguir clientes que compram regularmente de clientes que tendem a desaparecer]
- [ex.: decidir onde vale a pena investir em ofertas ou descontos]

O modelo **prevê**, não explica nem decide. Uma associação encontrada nos dados não prova uma relação de causa e efeito, e a decisão comercial continua a ser da empresa.

### Dados disponíveis

| Ficheiro | Conteúdo |
|---|---|
| `train.csv` | Ocasiões de compra históricas, com o target `repeat_purchase_90d` |
| `test.csv` | Ocasiões de compra posteriores, sem target |
| `clientes.csv` | Atributos ao nível do cliente, ligáveis através de `customer_id` (opcional) |
| `sample_submission.csv` | Estrutura de submissão e previsão de referência |

Cada linha representa uma ocasião de compra. O target é 1 se o mesmo cliente voltar a comprar nos 90 dias seguintes e 0 caso contrário. O `test.csv` é posterior ao treino no tempo, por isso o modelo tem de generalizar para o futuro.
